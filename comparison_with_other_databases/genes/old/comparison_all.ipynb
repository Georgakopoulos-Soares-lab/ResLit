```python
import pandas as pd

# Read and load resfinder_harmonized.tsv (all data)
resfinder_df = pd.read_csv('resfinder/resfinder_harmonized.tsv', sep='\t')
print(f"ResFinder rows: {len(resfinder_df)}")
resfinder_filtered = resfinder_df
 
# Read and load card_harmonized.tsv (all data)
card_df = pd.read_csv('card/card_harmonized.tsv', sep='\t')
print(f"CARD rows: {len(card_df)}")
card_filtered = card_df

# Read Reference Gene Catalog (all data)
ref_catalog_df = pd.read_csv('reference_gene_catalog/ReferenceGeneCatalog_amrGenes.csv')
print(f"Reference Gene Catalog rows: {len(ref_catalog_df)}")
ref_catalog_filtered = ref_catalog_df

print("\nDataframes ready for comparison (all data, no filtering): 'resfinder_filtered', 'card_filtered', and 'ref_catalog_filtered'")
```

```python
# Read genes.csv
genes_df = pd.read_csv('../normalised_genes_all_experimental_characterized_full.csv')
print(f"Total genes in genes.csv: {len(genes_df)}")

# Remove rows with NaN gene_name
genes_df_clean = genes_df[genes_df['gene_name'].notna()].reset_index(drop=True)
print(f"Genes after removing NaN gene_name: {len(genes_df_clean)}")

# Function to normalize gene names: remove special quote/prime characters and convert to lowercase
def normalize_gene_name(name):
    if pd.isna(name):
        return name
    name = str(name).lower()
    # Remove all prime/quote/apostrophe variants
    name = (name.replace('\u2032', '')   # ′ prime
                .replace('\u02b9', '')   # ʹ modifier letter prime
                .replace('\u2018', '')   # ' left single quote
                .replace('\u2019', '')   # ' right single quote / apostrophe
                .replace('\u0060', '')   # ` backtick
                .replace('\u00b4', '')   # ´ acute accent
                .replace("'", '')        # standard ASCII apostrophe
                .replace('"', ''))       # standard double quote just in case
    return name
                                                          

# Get unique gene names from each source (normalized and lowercase)
genes_resfinder = set(resfinder_filtered['gene_name'].dropna().apply(normalize_gene_name).unique())
genes_card = set(card_filtered['gene_name'].dropna().apply(normalize_gene_name).unique())
genes_ref_catalog = set(ref_catalog_filtered['gene_family'].dropna().apply(normalize_gene_name).unique())
genes_csv = set(genes_df_clean['gene_name'].apply(normalize_gene_name).unique())

print(f"\nUnique gene names in genes.csv (normalized): {len(genes_csv)}")
print(f"Unique gene names in resfinder_filtered (normalized): {len(genes_resfinder)}")
print(f"Unique gene names in card_filtered (normalized): {len(genes_card)}")
print(f"Unique gene families in ref_catalog_filtered (normalized): {len(genes_ref_catalog)}")

# Comparisons for each database
print("\n" + "="*70)
print("RESFINDER COMPARISON (Case-Insensitive, Normalized)")
print("="*70)
genes_csv_in_resfinder = genes_csv.intersection(genes_resfinder)
genes_resfinder_not_in_csv = genes_resfinder - genes_csv

print(f"Genes from genes.csv PRESENT in resfinder_filtered: {len(genes_csv_in_resfinder)}")
print(f"Genes from genes.csv NOT PRESENT in resfinder_filtered: {len(genes_csv - genes_resfinder)}")
print(f"Genes in resfinder_filtered but NOT in genes.csv: {len(genes_resfinder_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and resfinder (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_resfinder))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in resfinder (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_resfinder))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in resfinder but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_resfinder_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")

print("\n" + "="*70)
print("CARD COMPARISON (Case-Insensitive, Normalized)")
print("="*70)
genes_csv_in_card = genes_csv.intersection(genes_card)
genes_card_not_in_csv = genes_card - genes_csv

print(f"Genes from genes.csv PRESENT in card_filtered: {len(genes_csv_in_card)}")
print(f"Genes from genes.csv NOT PRESENT in card_filtered: {len(genes_csv - genes_card)}")
print(f"Genes in card_filtered but NOT in genes.csv: {len(genes_card_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and card (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_card))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in card (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_card))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in card but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_card_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")

print("\n" + "="*70)
print("REFERENCE GENE CATALOG COMPARISON (Case-Insensitive, Normalized)")
print("="*70)
genes_csv_in_ref_catalog = genes_csv.intersection(genes_ref_catalog)
genes_ref_catalog_not_in_csv = genes_ref_catalog - genes_csv

print(f"Genes from genes.csv PRESENT in ref_catalog_filtered: {len(genes_csv_in_ref_catalog)}")
print(f"Genes from genes.csv NOT PRESENT in ref_catalog_filtered: {len(genes_csv - genes_ref_catalog)}")
print(f"Genes in ref_catalog_filtered but NOT in genes.csv: {len(genes_ref_catalog_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and ref_catalog (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_ref_catalog))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in ref_catalog (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_ref_catalog))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in ref_catalog but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_ref_catalog_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")
```

```python
# Read genes.csv
genes_df = pd.read_csv('../normalised_genes_all_experimental_characterized_full.csv')
print(f"Total genes in genes.csv: {len(genes_df)}")

# Remove rows with NaN gene_name
genes_df_clean = genes_df[genes_df['gene_name'].notna()].reset_index(drop=True)
print(f"Genes after removing NaN gene_name: {len(genes_df_clean)}")

# Function to normalize gene names: remove everything after dash, remove special quote/prime characters, and convert to lowercase
def normalize_gene_name(name):
    if pd.isna(name):
        return name
    name = str(name).lower()
    # Remove everything after dash (keep only the base gene family)
    name = name.split('-')[0]
    # Remove all prime/quote/apostrophe variants
    name = (name.replace('\u2032', '')   # ′ prime
                .replace('\u02b9', '')   # ʹ modifier letter prime
                .replace('\u2018', '')   # ' left single quote
                .replace('\u2019', '')   # ' right single quote / apostrophe
                .replace('\u0060', '')   # ` backtick
                .replace('\u00b4', '')   # ´ acute accent
                .replace("'", '')        # standard ASCII apostrophe
                .replace('"', '')        # standard double quote just in case
                .replace(' ', ''))       # remove spaces
    return name
                                                          

# Get unique gene names from each source (normalized and lowercase)
genes_resfinder = set(resfinder_filtered['gene_name'].dropna().apply(normalize_gene_name).unique())
genes_card = set(card_filtered['gene_name'].dropna().apply(normalize_gene_name).unique())
genes_ref_catalog = set(ref_catalog_filtered['gene_family'].dropna().apply(normalize_gene_name).unique())
genes_csv = set(genes_df_clean['gene_name'].apply(normalize_gene_name).unique())

print(f"\nUnique gene names in genes.csv (normalized): {len(genes_csv)}")
print(f"Unique gene names in resfinder_filtered (normalized): {len(genes_resfinder)}")
print(f"Unique gene names in card_filtered (normalized): {len(genes_card)}")
print(f"Unique gene families in ref_catalog_filtered (normalized): {len(genes_ref_catalog)}")

# Comparisons for each database
print("\n" + "="*70)
print("RESFINDER COMPARISON (Base Gene Family Only)")
print("="*70)
genes_csv_in_resfinder = genes_csv.intersection(genes_resfinder)
genes_resfinder_not_in_csv = genes_resfinder - genes_csv

print(f"Genes from genes.csv PRESENT in resfinder_filtered: {len(genes_csv_in_resfinder)}")
print(f"Genes from genes.csv NOT PRESENT in resfinder_filtered: {len(genes_csv - genes_resfinder)}")
print(f"Genes in resfinder_filtered but NOT in genes.csv: {len(genes_resfinder_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and resfinder (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_resfinder))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in resfinder (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_resfinder))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in resfinder but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_resfinder_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")

print("\n" + "="*70)
print("CARD COMPARISON (Base Gene Family Only)")
print("="*70)
genes_csv_in_card = genes_csv.intersection(genes_card)
genes_card_not_in_csv = genes_card - genes_csv

print(f"Genes from genes.csv PRESENT in card_filtered: {len(genes_csv_in_card)}")
print(f"Genes from genes.csv NOT PRESENT in card_filtered: {len(genes_csv - genes_card)}")
print(f"Genes in card_filtered but NOT in genes.csv: {len(genes_card_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and card (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_card))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in card (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_card))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in card but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_card_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")

print("\n" + "="*70)
print("REFERENCE GENE CATALOG COMPARISON (Base Gene Family Only)")
print("="*70)
genes_csv_in_ref_catalog = genes_csv.intersection(genes_ref_catalog)
genes_ref_catalog_not_in_csv = genes_ref_catalog - genes_csv

print(f"Genes from genes.csv PRESENT in ref_catalog_filtered: {len(genes_csv_in_ref_catalog)}")
print(f"Genes from genes.csv NOT PRESENT in ref_catalog_filtered: {len(genes_csv - genes_ref_catalog)}")
print(f"Genes in ref_catalog_filtered but NOT in genes.csv: {len(genes_ref_catalog_not_in_csv)}")

print(f"\nSample of genes in both genes.csv and ref_catalog (first 10):")
for i, gene in enumerate(sorted(list(genes_csv_in_ref_catalog))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in genes.csv but NOT in ref_catalog (first 10):")
for i, gene in enumerate(sorted(list(genes_csv - genes_ref_catalog))[:10], 1):
    print(f"  {i}. {gene}")

print(f"\nSample of genes in ref_catalog but NOT in genes.csv (first 10):")
for i, gene in enumerate(sorted(list(genes_ref_catalog_not_in_csv))[:10], 1):
    print(f"  {i}. {gene}")
```
