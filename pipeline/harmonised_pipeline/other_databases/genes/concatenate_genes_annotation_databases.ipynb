```python
def split_and_keep_first(s):
    parts = s.split('_')
    if len(parts) > 1:
        return parts[0]
    else:
        return s, ''
```

```python
def split_and_keep_last(s):
    parts = s.split('_')
    if len(parts) > 1:
        return parts[-1]
    else:
        return s, ''
```

```python
file_path = './nr100_all_proteins_amr.fasta.clstr'  # Replace with your file path
same_proteins={}
with open(file_path, 'r') as file:
    content = file.read()
   # Split the content by '>Cluster' 
    clusters = content.split('>Cluster')[1:]  # Skip the first split part if it's empty
    
    # Regex pattern to find protein IDs
    pattern = re.compile(r'>(\S+)\.\.\.')
    pattern_first = re.compile(r'>(\S+)\.\.\. \*')
    pattern_second = re.compile(r'>(\S+)\.\.\. at')
    # List to store all found protein IDs
    protein_ids = []
    
    # Iterate over each cluster section
    for cluster in clusters:
        matches = pattern.findall(cluster)
        if len(matches)>1:
            first_protein=pattern_first.findall(cluster)
            second_proteins=pattern_second.findall(cluster)
            same_proteins[first_protein[0]]=second_proteins
            print(first_protein[0])
            print(second_proteins)
```

```python
result_all=result
result_all['ID'] = result['ID'].fillna('')
result_all = result_all.reset_index(drop=True)
result_all = result_all.reindex(list(range(len(result["ID"]))))
```

```python
for key, value in same_proteins.items():
    escaped_key = re.escape(key)
    row_1 = result_all.loc[result_all["ID"].str.contains(escaped_key, na=False)]
    
    if row_1.empty:
        continue
    
    row_1_index = row_1.index[0]
    
    # Create a regex pattern from the name list
    escaped_patterns = [re.escape(pattern) for pattern in value]
    combined_pattern = '|'.join(escaped_patterns)
    rows_other = result_all.loc[result_all["ID"].str.contains(combined_pattern, na=False)]
    
    if rows_other.empty:
        continue
    
    if len(value) == 1:
        for col in result_all.columns:
            if col != 'ID' and col != 'Database':
                if result_all.at[row_1_index, col] == '-':
                    result_all.at[row_1_index, col] = rows_other.iloc[0][col]
        result_all.at[row_1_index, "Database"] = result_all.at[row_1_index, "Database"] + "," + rows_other.iloc[0]["Database"]
        result_all.at[row_1_index, "subclass"] = result_all.at[row_1_index, "subclass"] + "," + rows_other.iloc[0]["subclass"]
        
    else:
        #print("Yeah")
        for col in result_all.columns:
            if col != 'ID' and col != 'Database':
                if result_all.at[row_1_index, col] == '-':
                    if rows_other.iloc[0][col] != '-':
                        result_all.at[row_1_index, col] = rows_other.iloc[0][col]
                    else:
                        result_all.at[row_1_index, col] = rows_other.iloc[1][col]
        result_all.at[row_1_index, "Database"] = result_all.at[row_1_index, "Database"] + "," + rows_other.iloc[0]["Database"] + "," + rows_other.iloc[1]["Database"]
        result_all.at[row_1_index, "subclass"] = result_all.at[row_1_index, "subclass"] + "," + rows_other.iloc[0]["subclass"] + "," + rows_other.iloc[1]["subclass"]
    
 
```

```python
# Function to process each row
def process_row(row):
    # Split the subclass string into individual words using , ; /
    if pd.isna(row["subclass"]) or row["subclass"] == "NaN":
        return "NaN"  # Return "NaN" string if subclass is NaN
    else: 
        words = [word.strip() for word in re.split(r'[;,/]', row["subclass"]) if word.strip()]

        # Convert to uppercase and keep only unique words, excluding "-"
        unique_words = set(word.upper() for word in words if word != "-")

        # Join the unique words into a single string with commas
        return ", ".join(unique_words)


# Apply the function to each row in the "subclass" column
result_all["subclass"] = result_all.apply(lambda row: process_row(row), axis=1)
```
